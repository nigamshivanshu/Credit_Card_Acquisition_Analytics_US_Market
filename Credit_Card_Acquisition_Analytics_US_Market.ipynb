# Synthetic Data Generation

---

## 1. Purpose of Synthetic Data

To simulate real-world credit card acquisition analytics, this project uses large-scale synthetic data designed to reflect realistic customer behavior, acquisition funnels, and revenue patterns.

The dataset intentionally includes noise, missing values, and inconsistencies to replicate the data quality challenges commonly encountered in production analytics environments.

---

## 2. Data Volume Overview

| Table | Approximate Rows |
|-----|------------------|

This scale enables performance analysis, aggregation challenges, and realistic KPI behavior.

## 3. Realism & Noise Design

The data intentionally includes:
- Rejected applications
- Non-activated approved customers
- Skewed transaction amounts
- Missing demographic values
- Delayed activation behavior
- Channel-level performance differences

These characteristics ensure the analysis reflects real-world ambiguity rather than idealized outcomes.

---


```python
# Import Libraries
import numpy as np
import pandas as pd

np.random.seed(42)
```

```python
# Generate Customers Table

n_customers = 150_000

customers = pd.DataFrame({
    "customer_id": np.arange(1, n_customers + 1),
    "acquisition_channel": np.random.choice(
        ["Digital", "Referral", "Partner", "Organic"],
        size=n_customers,
        p=[0.45, 0.25, 0.20, 0.10]
    ),
    "segment": np.random.choice(
        ["Mass", "Affluent", "SME"],
        size=n_customers,
        p=[0.65, 0.25, 0.10]
    ),
    "state": np.random.choice(
        ["CA", "NY", "TX", "FL", "IL", "PA", "OH", "GA", "NC", "MI"],
        size=n_customers
    ),
    "signup_date": pd.to_datetime("2024-01-01") +
        pd.to_timedelta(np.random.randint(0, 180, size=n_customers), unit="D")
})

# Introduce missing values (realistic noise)
mask = np.random.rand(n_customers) < 0.03
customers.loc[mask, "segment"] = None

customers.head()
```

```python
# Generate Applications Table

applications = customers[["customer_id", "signup_date"]].copy()
applications["application_id"] = np.arange(1, n_customers + 1)

applications["application_status"] = np.random.choice(
    ["Approved", "Rejected"],
    size=n_customers,
    p=[0.70, 0.30]
)

applications["approval_date"] = pd.NaT

approved_mask = applications["application_status"] == "Approved"

applications.loc[approved_mask, "approval_date"] = (
    applications.loc[approved_mask, "signup_date"]
    + pd.to_timedelta(
        np.random.randint(1, 7, size=approved_mask.sum()),
        unit="D"
    )
)


applications.rename(columns={"signup_date": "application_date"}, inplace=True)
applications.head()
```

```python
# ACTIVATIONS TABLE

approved = applications[applications["application_status"] == "Approved"].copy()
approved = approved.reset_index(drop=True)

activation_prob = 0.75
activation_flags = np.random.rand(len(approved)) < activation_prob

activations = pd.DataFrame({
    "customer_id": approved["customer_id"],
    "activation_status": np.where(activation_flags, "Activated", "Not Activated"),
    "activation_date": pd.NaT
})

# Assign activation dates ONLY for activated customers
activated_idx = activations.index[activations["activation_status"] == "Activated"]

activations.loc[activated_idx, "activation_date"] = (
    approved.loc[activated_idx, "approval_date"]
    + pd.to_timedelta(
        np.random.randint(1, 14, size=len(activated_idx)),
        unit="D"
    )
)

activations.head()
```

```python
# TRANSACTIONS TABLE

active_customers = activations.loc[
    activations["activation_status"] == "Activated", "customer_id"
]

n_transactions = 600_000

transactions = pd.DataFrame({
    "transaction_id": np.arange(1, n_transactions + 1),
    "customer_id": np.random.choice(active_customers, size=n_transactions),
    "transaction_date": pd.to_datetime("2024-01-01")
        + pd.to_timedelta(np.random.randint(0, 180, size=n_transactions), unit="D"),
    "transaction_amount": np.round(
        np.random.lognormal(mean=3.5, sigma=1.0, size=n_transactions), 2
    ),
    "merchant_category": np.random.choice(
        ["Retail", "Travel", "Dining", "Fuel", "E-commerce"],
        size=n_transactions,
        p=[0.30, 0.25, 0.20, 0.15, 0.10]
    )
})

# Inject realistic noise: zero & extreme transactions
noise_mask = np.random.rand(n_transactions) < 0.01
transactions.loc[noise_mask, "transaction_amount"] *= np.random.choice([0, 5], size=noise_mask.sum())

transactions.head()
```

```python
# Generate Revenue Table

# REVENUE TABLE (derived from transactions)

revenue = transactions.sample(frac=0.5, random_state=42).copy()

revenue["revenue_date"] = revenue["transaction_date"]

# Revenue as a % of transaction value (interchange + fees simulation)
revenue["revenue_amount"] = np.round(
    revenue["transaction_amount"] * np.random.uniform(0.01, 0.03, size=len(revenue)),
    2
)

# Introduce missing revenue values (real-world noise)
missing_rev_mask = np.random.rand(len(revenue)) < 0.02
revenue.loc[missing_rev_mask, "revenue_amount"] = np.nan

revenue = revenue[["customer_id", "revenue_date", "revenue_amount"]]

revenue.head()
```

```python
# MARKETING SPEND TABLE

dates = pd.date_range(start="2024-01-01", periods=180, freq="D")
channels = ["Digital", "Referral", "Partner", "Organic"]

marketing_spend = pd.DataFrame({
    "channel": np.repeat(channels, len(dates)),
    "spend_date": np.tile(dates, len(channels)),
    "spend_amount": np.round(
        np.random.uniform(5_000, 50_000, size=len(dates) * len(channels)),
        2
    )
})

# Channel-specific spend bias (more realistic)
marketing_spend.loc[marketing_spend["channel"] == "Digital", "spend_amount"] *= 1.4
marketing_spend.loc[marketing_spend["channel"] == "Organic", "spend_amount"] *= 0.4

marketing_spend.head()
```

# STEP 5: Acquisition Funnel Analytics

---

## 1. Funnel Definition

The acquisition funnel represents the customer journey from initial application through approval and eventual card activation.

The funnel stages analyzed are:
1. Applications submitted
2. Applications approved
3. Cards activated

Analyzing conversion rates at each stage helps identify operational bottlenecks and optimization opportunities across acquisition channels.

---


```python
# Merge core funnel tables
funnel_df = (
    customers[["customer_id", "acquisition_channel"]]
    .merge(applications[["customer_id", "application_status"]], on="customer_id", how="left")
    .merge(activations[["customer_id", "activation_status"]], on="customer_id", how="left")
)

funnel_df.head()
```

```python
funnel_df["applied"] = 1

funnel_df["approved"] = (funnel_df["application_status"] == "Approved").astype(int)

funnel_df["activated"] = (funnel_df["activation_status"] == "Activated").astype(int)

funnel_df.head()
```

```python
# OVERALL FUNNEL METRICS
funnel_summary = funnel_df.agg({
    "applied": "sum",
    "approved": "sum",
    "activated": "sum"
}).to_frame(name="count")

funnel_summary["conversion_rate"] = funnel_summary["count"] / funnel_summary.loc["applied", "count"]

funnel_summary
```

```python
# FUNNEL BY ACQUISITION CHANNEL
channel_funnel = (
    funnel_df
    .groupby("acquisition_channel")
    .agg(
        applications=("applied", "sum"),
        approvals=("approved", "sum"),
        activations=("activated", "sum")
    )
)

channel_funnel["approval_rate"] = channel_funnel["approvals"] / channel_funnel["applications"]
channel_funnel["activation_rate"] = channel_funnel["activations"] / channel_funnel["approvals"]

channel_funnel = channel_funnel.sort_values("activation_rate", ascending=False)

channel_funnel
```

```python
# FUNNEL DROP-OFF ANALYSIS

channel_funnel["approval_dropoff"] = 1 - channel_funnel["approval_rate"]
channel_funnel["activation_dropoff"] = 1 - channel_funnel["activation_rate"]

channel_funnel[[
    "approval_rate",
    "activation_rate",
    "approval_dropoff",
    "activation_dropoff"
]]
```

## Acquisition Funnel Performance — Business Interpretation

### Overall Funnel Health

The end-to-end acquisition funnel shows that approximately **70% of applications are approved**, while only **~53% of applicants ultimately activate** their cards.  
This indicates that while credit approval performance is relatively strong, **activation represents the largest point of customer drop-off** in the acquisition lifecycle.

The gap between approval and activation suggests onboarding friction or reduced customer intent post-approval, making activation optimization a high-impact opportunity for improving acquisition efficiency.

---

### Channel-Level Funnel Performance

Across acquisition channels, approval rates remain broadly consistent, ranging between **~69% and ~70%**, indicating that credit risk and underwriting criteria are applied uniformly across channels.

However, meaningful differences emerge when analyzing activation performance:

- **Partner and Digital channels** demonstrate slightly stronger activation rates (~75%), indicating higher customer intent or better onboarding effectiveness.
- **Referral and Organic channels**, while delivering steady application volumes, exhibit marginally lower activation rates, contributing to higher post-approval drop-offs.

Although the absolute differences appear modest, even small improvements in activation rates can translate into a significant increase in activated customers at scale.

---

### Funnel Drop-Off Insights

Activation drop-off rates (~24–25%) are consistently higher than approval drop-offs across all channels.  
This highlights that **customer engagement after approval is a more critical bottleneck than application screening**.

Potential drivers may include:
- Delays in card delivery or activation communication
- Suboptimal onboarding experiences
- Lower immediate perceived value post-approval

---

### Strategic Implications

- **Prioritize activation optimization** as the primary lever for improving acquisition ROI, rather than increasing top-of-funnel volume.
- **Sustain investment in Partner and Digital channels**, which demonstrate relatively stronger end-to-end funnel efficiency.
- **Target onboarding improvements** for Referral and Organic channels to reduce post-approval attrition.
- Introduce channel-specific activation interventions such as guided activation journeys, incentives, or personalized communication.

---

### Executive Takeaway

Improving activation rates by even **2–3 percentage points** would materially increase the number of revenue-generating customers without additional acquisition spend.  
As a result, activation-focused initiatives represent one of the highest ROI opportunities within the current acquisition strategy.


# STEP 6: Marketing Spend, CAC & Channel ROI Analysis

---

## 1. Objective

This analysis evaluates the efficiency of marketing spend across acquisition channels by linking channel-level investment to customer acquisition outcomes.

The goal is to understand:
- How much it costs to acquire an activated customer
- Which channels deliver the best balance of volume, quality, and cost efficiency
- Where marketing investments can be optimized to improve ROI

---


```python
# Aggregate Marketing Spend by Channel

channel_spend = (
    marketing_spend
    .groupby("channel", as_index=False)
    .agg(total_spend=("spend_amount", "sum"))
)

channel_spend
```

```python
channel_metrics = (
    channel_funnel
    .reset_index()
    .merge(channel_spend, left_on="acquisition_channel", right_on="channel", how="left")
)

channel_metrics
```

```python
# CAC Calculations

channel_metrics["CAC"] = (
    channel_metrics["total_spend"] / channel_metrics["activations"]
)

channel_metrics = channel_metrics.sort_values("CAC")

channel_metrics[[
    "acquisition_channel",
    "applications",
    "approvals",
    "activations",
    "approval_rate",
    "activation_rate",
    "total_spend",
    "CAC"
]]
```

```python
# CAC VS FUNNEL QUALITY

channel_metrics["cost_per_application"] = (
    channel_metrics["total_spend"] / channel_metrics["applications"]
)

channel_metrics["cost_per_approval"] = (
    channel_metrics["total_spend"] / channel_metrics["approvals"]
)

channel_metrics[[
    "acquisition_channel",
    "cost_per_application",
    "cost_per_approval",
    "CAC"
]]
```

## Marketing Spend & CAC — Business Interpretation

### Overall Spend Allocation

Marketing spend is unevenly distributed across acquisition channels, with **Digital** receiving the highest investment, followed by **Partner** and **Referral** channels.  
Despite lower total spend, **Organic** continues to contribute a meaningful volume of activated customers, indicating strong inherent intent but limited scalability.

---

## Customer Acquisition Cost (CAC) Comparison

Customer Acquisition Cost (CAC) varies significantly across acquisition channels:

- **Digital** delivers the lowest CAC (~198), driven by strong activation performance at scale.
- **Referral** and **Organic** channels exhibit moderate CAC (~242–247), reflecting stable funnel performance but lower cost efficiency.
- **Partner** channel shows the highest CAC (~329), despite comparable approval and activation rates.

This analysis highlights that cost efficiency is not solely driven by funnel conversion rates, but also by the level of marketing investment required to acquire customers.


---

### Cost vs Funnel Efficiency Trade-Off

While Partner and Digital channels demonstrate similar activation rates (~75%), their cost structures differ materially.  
The Partner channel requires significantly higher spend per application and per approval, reducing overall acquisition efficiency.

Digital, by contrast, achieves a more favorable balance between **scale, activation efficiency, and cost**, making it the most efficient growth driver in the current mix.

---

### Strategic Implications

- **Prioritize incremental investment in Digital acquisition**, where lower CAC and strong activation rates maximize return on spend.
- **Maintain Referral and Organic channels** as complementary acquisition sources, while exploring cost optimization opportunities.
- **Reassess Partner channel strategy**, focusing on improving targeting efficiency, renegotiating partnership costs, or refining partner selection criteria.

---

### Executive Recommendation

Reallocating a portion of Partner channel spend toward lower-CAC channels could increase the number of activated customers without increasing overall marketing budget.  
A CAC-driven acquisition strategy enables sustainable growth by balancing volume, customer quality, and cost efficiency.


# STEP 7: Revenue, Customer Quality & Early LTV Analysis

---

## 1. Objective

This analysis evaluates the quality of acquired customers by linking acquisition channels and customer segments to early-stage revenue generation.

Rather than focusing solely on acquisition cost, the objective is to understand which channels and segments deliver customers with stronger revenue potential and long-term value signals.

---


```python
# Aggregate Revenue at Customer Level

customer_revenue = (
    revenue_clean
    .groupby("customer_id", as_index=False)
    .agg(
        total_revenue=("revenue_amount", "sum"),
        revenue_events=("revenue_amount", "count")
    )
)

customer_revenue.head()
```

```python
customer_value = (
    customers[["customer_id", "acquisition_channel", "segment"]]
    .merge(customer_revenue, on="customer_id", how="left")
)

customer_value["total_revenue"] = customer_value["total_revenue"].fillna(0)
customer_value["revenue_events"] = customer_value["revenue_events"].fillna(0)

customer_value.head()
```

```python
# REVENUE QUALITY BY CHANNEL
channel_revenue_quality = (
    customer_value
    .groupby("acquisition_channel")
    .agg(
        customers=("customer_id", "nunique"),
        avg_revenue_per_customer=("total_revenue", "mean"),
        median_revenue=("total_revenue", "median"),
        pct_revenue_generating=("total_revenue", lambda x: (x > 0).mean())
    )
    .sort_values("avg_revenue_per_customer", ascending=False)
)

channel_revenue_quality
```

```python
# EARLY LTV PROXY (PRACTICAL & REALISTIC)

# Early LTV proxy = avg revenue per customer * activation rate
early_ltv = (
    channel_revenue_quality
    .merge(
        channel_funnel[["activation_rate"]],
        left_index=True,
        right_index=True
    )
)

early_ltv["early_LTV_proxy"] = (
    early_ltv["avg_revenue_per_customer"] * early_ltv["activation_rate"]
)

early_ltv.sort_values("early_LTV_proxy", ascending=False)
```

```python
# SEGMENT-LEVEL QUALITY ANALYSIS

segment_quality = (
    customer_value
    .groupby("segment")
    .agg(
        customers=("customer_id", "nunique"),
        avg_revenue=("total_revenue", "mean"),
        median_revenue=("total_revenue", "median")
    )
    .sort_values("avg_revenue", ascending=False)
)

segment_quality
```

While many customers generate modest early-stage revenue, a meaningful portion show repeated revenue events, signaling engagement and long-term value potential.

Average revenue per customer remains relatively consistent across acquisition channels, ranging from approximately **$2.15 to $2.18**.

- **Digital channel**, while highly cost-efficient and scalable, shows slightly lower early revenue per customer compared to other channels.

These patterns suggest that volume-efficient channels do not always deliver the highest-value customers.

---

### Early LTV Proxy Insights

By combining average revenue per customer with activation rates, an early LTV proxy highlights meaningful differences in value contribution across channels:
